```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("/content/student_data (2).csv")
df.columns = df.columns.str.strip()

sns.set_theme(style="whitegrid")

# Check available columns
print("Dataset size:", df.shape)
print("Available columns:", df.columns.tolist())

# 1. BAR CHART: Average final grade by school
if "G3" in df.columns and "school" in df.columns:
    plt.figure(figsize=(7, 5))
    sns.barplot(data=df, x="school", y="G3", errorbar=None)
    plt.title("Average Final Grade by School")
    plt.xlabel("School")
    plt.ylabel("Average Final Grade (G3)")
    plt.tight_layout()
    plt.savefig("bar_chart.png", dpi=300)
    plt.show()
else:
    print("Bar chart: Required columns not found.")

# 2. LINE CHART: Final grade by age
if "G3" in df.columns and "age" in df.columns:
    age_grade = df.groupby("age")["G3"].mean().reset_index()

    plt.figure(figsize=(7, 5))
    sns.lineplot(data=age_grade, x="age", y="G3", marker="o")
    plt.title("Average Final Grade by Age")
    plt.xlabel("Age")
    plt.ylabel("Average Final Grade (G3)")
    plt.tight_layout()
    plt.savefig("line_chart.png", dpi=300)
    plt.show()
else:
    print("Line chart: Required columns not found.")

# 3. SCATTER PLOT: First-period vs final grade
if all(c in df.columns for c in ["G1", "G3"]):
    plt.figure(figsize=(7, 5))
    sns.scatterplot(data=df, x="G1", y="G3")
    plt.title("First-Period Grade vs Final Grade")
    plt.xlabel("First-Period Grade (G1)")
    plt.ylabel("Final Grade (G3)")
    plt.tight_layout()
    plt.savefig("scatter_plot.png", dpi=300)
    plt.show()
else:
    print("Scatter plot: G1 or G3 column not found.")

# 4. HEATMAP: Correlation among numerical columns
numeric_df = df.select_dtypes(include="number")

if len(numeric_df.columns) >= 2:
    plt.figure(figsize=(12, 9))
    sns.heatmap(
        numeric_df.corr(),
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )
    plt.title("Correlation Heatmap of Student Data")
    plt.tight_layout()
    plt.savefig("heatmap.png", dpi=300)
    plt.show()
else:
    print("Heatmap: Not enough numerical columns.")

print("All available visualizations generated.")
```

```python
from google.colab import files
import os
import zipfile

chart_files = [
    "bar_chart.png",
    "line_chart.png",
    "scatter_plot.png",
    "heatmap.png"
]

with zipfile.ZipFile("Week3_Charts.zip", "w") as zipf:
    for filename in chart_files:
        if os.path.exists(filename):
            zipf.write(filename)

files.download("Week3_Charts.zip")
```
